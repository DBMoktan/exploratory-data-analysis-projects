```python
# Select only numeric columns (assuming all non-numeric columns should be excluded)
numeric_columns = df.select_dtypes(include=[np.number])

# Calculate the correlation matrix
corr_df = numeric_columns.corr(method='pearson')

# Display correlation
corr_df
```

```python
plt.figure(figsize=(12, 6))
heatmap = sns.heatmap(corr_df, annot=True, fmt='.1g', vmin=-1, center=0, cmap='inferno', linewidths=1, linecolor='black')
heatmap.set_title("Correlation Heatmap Between Variables")
```

```python
# Remove rows with invalid date values
df = df[df['release_date'].str.match(r'\d{4}-\d{2}-\d{2}')]
```

```python
# Set the release date column as the index
df.set_index('release_date', inplace=True)
df.index = pd.to_datetime(df.index)

print(df)
```

```python
df.index
```

```python
df.index.get_level_values('release_date')
```

```python
# Create a new column called years from release date column
df['dates'] = df.index.get_level_values('release_date')
df.dates = pd.to_datetime(df.dates)
years = df.dates.dt.year

years
```

```python
# Distribution plot (bar plot) for total number of song per year
plt.figure(figsize=(10,6))
sns.displot(years, discrete=True, aspect=2, height=5, kind='hist').set_titles('Total numbers of songs per year')
```

```python
# Convert duration in milliseconds to seconds
df['duration'] = df['duration_ms'].apply(lambda x: round(x/1000))
df.drop('duration_ms', inplace = True, axis = 1)
```

```python
# Duration of the songs over the years
total_dr = df.duration

print(total_dr)
```

```python
# Barplot for Year vs Duration
fig, ax = plt.subplots(figsize=(18, 8))
sns.barplot(x=years, y=total_dr, ax=ax)
ax.set(title='Year vs Duration')
plt.xticks(rotation=90)
ax.set_xlabel('Year')  # Add x-axis label
```

```python
# Line plot for average duration of the songs over the years
total_dr = df.duration

print(total_dr)
```

```python
plt.figure(figsize=(10, 5))
sns.lineplot(x=years, y=total_dr)
plt.title('Year vs Duration')
plt.xlabel('Year')
plt.show()
```
